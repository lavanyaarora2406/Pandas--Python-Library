MISSING DATA

FINDING MISSING DATA 

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
# np.nan is used to reperesent null values .. 
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

REMOVINg missing data 

```python
df
```

```python
# removing missing data is based on the rows .. 
```

```python
df.dropna()
```

```python
df.dropna(thresh = 3) # this will remove rows which have less than 3 non null values
```

FILLING THE MISSING DATA 

```python
df
```

```python
df.fillna(0) # this will fill the missing values with 0
```

Filling columns  with different  values  

```python
values = {'A': 0, 'B': 1, 'C': 2, 'D': 3}
df.fillna(value = values) # this will fill the missing values with the specified values for each column
```
