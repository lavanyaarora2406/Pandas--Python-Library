MERGING JOINING CONCTENATION        

```python
import numpy as np 
import pandas as pd 
```

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})

```

Merging 2 Dataframes 

```python
employees
```

```python
salaries
```

```python
pd.merge(employees , salaries , on = "employee_id", how = "inner") # Inner join: returns only the rows with matching employee_id in both DataFrames.
```

```python
pd.merge(employees, salaries, on='employee_id', how='outer') # Outer join: returns all rows from both DataFrames, filling in NaN for missing matches.
```

```python
pd.merge(employees, salaries, on='employee_id', how='left') # Left join: returns all rows from the left DataFrame (employees) and matching rows from the right DataFrame (salaries). Non-matching rows will have NaN for salary and bonus.
```

```python
pd.merge(employees, salaries, on='employee_id', how='right') # Right join: returns all rows from the right DataFrame (salaries) and matching rows from the left DataFrame (employees). Non-matching rows will have NaN for name and department.
```

CONCATENATION OF 2 DATAFRAMES 

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1 , df2 ])
```

```python
pd.concat([df1, df2], axis=1)  # Concatenates along rows (default behavior)  (Based on the rows .. )

# if axis = 0 it is based on the columns and if axis = 1 it is based on the rows.... 
```

JOINING 2 DATAFRAMES 

```python

df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])

```

```python
df1
```

```python
df2
```

```python
df1.join(df2) 
```
